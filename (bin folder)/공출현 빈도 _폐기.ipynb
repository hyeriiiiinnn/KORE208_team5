```python
import json
import re
from kiwipiepy import Kiwi
from collections import Counter

# Kiwi 형태소 분석기 초기화
kiwi = Kiwi()

# JSON 파일 불러오기 함수
def load_articles(*filepaths):
    articles = []
    for path in filepaths:
        with open(path, encoding='utf-8') as f:
            articles.extend(json.load(f))
    return articles

# 불용어 리스트
KOREAN_STOPWORDS = {
    '그', '이', '것', '저', '등', '수', '때', '좀', '더', '등등', '저희', '우리', '같이', '자신',
    '있다', '없다', '되다', '하다', '위해', '대한', '그리고', '그러나', '하지만', '또는',
    '이런', '그런', '저런', '아니다', '이다', '입니다', '하는', '했다', '했다가', '하며',
    '하고', '였다', '에서', '으로', '에게', '보다', '까지', '부터', '만큼', '조차',
    '같은', '모든', '각', '어떤', '다른', '이후', '앞서', '최근', '현재', '동안', '경우',
    '또한', '뿐만', '이미', '모두', '가장', '때문', '따라', '즉', '즉시', '다시', '그때',
    '이제', '계속', '그동안', '중에', '중에서', '이른바', '바로', '정도', '정확히',
    '기자', 'ᆫ', '후보', '이', '하', '“', '”', '있', '것', '(', ')'
}

EXCLUDE_TAGS = {
    # 조사
    'JKS', 'JKC', 'JKG', 'JKO', 'JKB', 'JKV', 'JKQ', 'JX', 'JC',
    # 어미
    'EP', 'EF', 'EC', 'ETN', 'ETM',
    # 파생 접사
    'XSV', 'XSA',
    # 숫자/기호/문장부호
    'SN', 'SF', 'SP', 'SS', 'SE', 'SW',
    # 한자/외래어/미확정
    'SH', 'SL', 'UNKNOWN'
}

# 후보 + 키워드 동시 등장 기사에서 단어 n개 추출 함수
def get_cooccurring_words(articles, candidate_name, keyword,
                           top_n=20,
                           stopwords=None):
    counter = Counter()

    for article in articles:
        content = article.get("content", "")
        content_clean = re.sub(r'\s+', ' ', content)

        if candidate_name in content_clean and keyword in content_clean:
            # 형태소 분석 → 품사 필터링 먼저 적용
            tokens = [
                word.form for word in kiwi.tokenize(content_clean)
                if word.tag not in EXCLUDE_TAGS
            ]

            # 불용어 제거
            if stopwords:
                tokens = [t for t in tokens if t not in stopwords]

            counter.update(tokens)

    return counter.most_common(top_n)


articles = load_articles("이재명.json", "김문수.json", "이준석.json")
```

```python
results = get_cooccurring_words(
    articles,
    candidate_name="김문수",
    keyword="단일",
    top_n=15,
    stopwords=KOREAN_STOPWORDS
)

# 결과 출력
for word, count in results:
    print(f"{word}: {count}회")
```
